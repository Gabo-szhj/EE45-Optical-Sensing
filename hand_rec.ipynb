```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

# Load Sign Language MNIST (Kaggle format: CSV with labels in first column)
train_df = pd.read_csv('sign_mnist_train.csv')
test_df = pd.read_csv('sign_mnist_test.csv')

# Extract labels and images
y_train = train_df['label'].values
y_test = test_df['label'].values
X_train = train_df.drop('label', axis=1).values.reshape(-1, 28, 28, 1)
X_test = test_df.drop('label', axis=1).values.reshape(-1, 28, 28, 1)

# Normalize and convert to float32
X_train = X_train / 255.0
X_test = X_test / 255.0

# Downscale 28x28 to 7x7 (optional: skip if you want to pre-train on 28x28)
def downscale(images):
    return tf.image.resize(images, (7, 7), method='area').numpy()

X_train_small = downscale(X_train)  # (27455, 7, 7, 1)
X_test_small = downscale(X_test)    # (7172, 7, 7, 1)
```
